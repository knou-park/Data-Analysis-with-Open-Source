```python
import requests

url = 'https://apis.data.go.kr/B552845/perDay/price'
api_key = '45d42ae65f63d1e55f397718f9485dabc76dce85a1233dffb2327f32d095b35d'

vrty = ['01', '02', '03', '06']                 # 배추 품종코드
total_data = []                                 # 2015년부터 2024년까지의 데이터 누적용 리스트

for exmn in range(2015, 2025) :                 # 년도별 데이터 누적용 반복문

    gte = f'{exmn}0101'                         # 각 년도별 검색 시작일
    lte = f'{exmn}1231'                         # 각 년도별 검색 종료일

    for vr in vrty :                            # 배추 품종 데이터 누적용 반복문

        params = {
            'serviceKey': api_key,              # 인증키
            'returnType': 'JSON',               # 결과 유형       default:JSON
            'pageNo': '1',                      # 페이지 번호     default: 1
            'numOfRows': '200',                 # 페이지 결과 수  default: 10, max:1000
            'cond[exmn_ymd::LTE]': lte,         # 조사일자가 검색값과 일치하거나 작은 데이터를 검색  format:YYYYMMDD
            'cond[exmn_ymd::GTE]': gte,         # 조사일자가 검색값과 일치하거나 큰 데이터를 검색    format:YYYYMMDD
            'cond[se_cd::EQ]': '02',            # 검색값과 구분코드가 일치하는 데이터를 검색
            'cond[ctgry_cd::EQ]': '200',        # 검색값과 부류코드가 일치하는 데이터를 검색
            'cond[item_cd::EQ]': '211',         # 검색값과 품목코드가 일치하는 데이터를 검색
            'cond[vrty_cd::EQ]': vr,            # 검색값과 품종코드가 일치하는 데이터를 검색
            'cond[grd_cd::EQ]': '04',           # 검색값과 등급코드가 일치하는 데이터를 검색
            'cond[sgg_cd::EQ]': '1101',         # 검색값과 시군구코드가 일치하는 데이터를 검색
            'cond[mrkt_cd::EQ]': '0110211',     # 검색값과 시장코드가 일치하는 데이터를 검색
            #'selectable': ''                   # 응답으로 받고 싶은 컬럼명을 선택하여 검색(구분자','로 다중선택)  e.g.) item_cd,vrty_cd
}

# 데이터 수집
        response = requests.get(url, params=params)

# requests로부터 response, body, items, item 항목 읽기
        data = response.json()['response']['body']['items']['item']
        total_data.extend(data)

# 호출 성공/실패 출력
        print(f"{exmn}년 품종코드: {vr}", response.json())
```
